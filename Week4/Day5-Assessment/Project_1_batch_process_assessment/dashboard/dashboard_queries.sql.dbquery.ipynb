```sql
-- Databricks SQL Dashboard queries
-- KPI 1: Total trades
SELECT COUNT(*) AS total_trades FROM week4_batch.trades.silver_trades;

-- KPI 2: Gross trade value
SELECT ROUND(SUM(trade_value), 2) AS gross_trade_value FROM week4_batch.trades.silver_trades;

-- KPI 3: Net flow
SELECT ROUND(
  SUM(CASE WHEN side = 'BUY' THEN trade_value ELSE -trade_value END), 2
) AS net_flow FROM week4_batch.trades.silver_trades;

-- KPI 4: Quarantined records
SELECT COUNT(*) AS quarantined_rows FROM week4_batch.trades.quarantine_trades;

-- Chart: Daily gross value
SELECT trade_date, ROUND(SUM(trade_value), 2) AS gross_value
FROM week4_batch.trades.silver_trades
GROUP BY trade_date ORDER BY trade_date;

-- Chart: Symbol activity
SELECT symbol, COUNT(*) AS trades, ROUND(SUM(trade_value), 2) AS gross_value
FROM week4_batch.trades.silver_trades
GROUP BY symbol ORDER BY gross_value DESC LIMIT 10;

-- Chart: Buy vs Sell
SELECT side, COUNT(*) AS trades, ROUND(SUM(trade_value), 2) AS trade_value
FROM week4_batch.trades.silver_trades
GROUP BY side;

-- Chart: Segment
SELECT segment, COUNT(*) AS trades, ROUND(SUM(trade_value), 2) AS gross_value
FROM week4_batch.trades.silver_trades
GROUP BY segment ORDER BY gross_value DESC;
```
