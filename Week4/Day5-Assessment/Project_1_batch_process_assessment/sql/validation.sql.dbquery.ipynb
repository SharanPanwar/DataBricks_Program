```sql
CREATE SCHEMA IF NOT EXISTS week4_batch.trades;

CREATE OR REPLACE VIEW week4_batch.trades.vw_trade_quality AS
SELECT
  COUNT(*) AS total_rows,
  SUM(CASE WHEN trade_id IS NULL THEN 1 ELSE 0 END) AS null_trade_id,
  SUM(CASE WHEN quantity <= 0 THEN 1 ELSE 0 END) AS invalid_quantity,
  SUM(CASE WHEN price <= 0 THEN 1 ELSE 0 END) AS invalid_price
FROM week4_batch.trades.silver_trades;

SELECT * FROM week4_batch.trades.gold_daily_position
ORDER BY trade_date, symbol;

-- Delta maintenance / inspection examples
DESCRIBE HISTORY week4_batch.trades.silver_trades;
DESCRIBE DETAIL week4_batch.trades.silver_trades;
```
